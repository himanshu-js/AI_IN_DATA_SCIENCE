# SEABORN



* Seaborn is a Python data visualization library built on top of Matplotlib and Numpy

*  Highly preferred for quick exploratory data analysis and advanced statistical plotting.
* Work with pandas dataframes
*   Comes with built-in datasets like iris, tips, etc.
* Provides statistical plots such as boxplots, violin plots, swarm plots, etc.
* Handles categorical data visualization better than Matplotlib.
* Supports aesthetic customization (themes, color palettes, styles).
* Simplifies working with DataFrames by auto-labeling axes.

## KDE(Kernal Density Estimation)



* Density of Probability of Values: KDE visualizes how data points are distributed over a continuous interval. It shows where values are concentrated.
* Uses Probability Density Function (PDF): The main goal of KDE is to estimate the underlying PDF of the dataset, ensuring the total area under the estimated curve equals 1.
* Peaks & Values: The peaks of a KDE plot represent the modes (the most common values or highest density regions) in your data.Shapes: It helps identify the overall structure of the data, such as whether it is unimodal (one peak), bimodal (two peaks), or multimodal.
* Skewness: You can easily see if the data is symmetric, left-skewed (long tail on the left), or right-skewed (long tail on the right).

### DETACTING OUTLIERS

```python
# taking the data

data = [9,5,6,7,40,11,4,3]

# sorting the data using the numpy

data_sort = np.sort(data)
data_sort
```

```python
# finding the Q1 = 25%,Q2 = 50%,Q3 = 75%

Q1 = np.percentile(data_sort,25,method='midpoint')

Q2 = np.percentile(data_sort,50,method='midpoint')

Q3 = np.percentile(data_sort,75,method='midpoint')

# IQR

IQR = Q3-Q1

print("Q1 25 percentile of the given data:",Q1)
print("Q2 50 percentile of the given data:",Q2)
print("Q3 75 percentile of the given data:",Q3)


print("IQR:",IQR)

```

```python
# finding the lower and upper limit

lower_limit = Q1 - 1.5*IQR
upper_limit = Q3 + 1.5*IQR

print("Lower limit:",lower_limit)
print("Upper limit:",upper_limit)
```

```python
# identifying the outliers

outliers = []

for x in data_sort:
  if ((x > upper_limit) or ( x < lower_limit)):
    outliers.append(x)
print('outlier in the dataset is:',outliers)
```

```python
sns.boxplot(data_sort)
plt.show()
```

## VIOLIN PLOT
* A violin plot combines a box plot with a density plot, showing both summary stats and distribution shape.

* Comparing distributions more deeply than boxplot.
* Helpful for detecting multimodal distributions.

* Shows both summary statistics and data distribution.
* Easier to see differences in distribution shapes.

* Can be harder to interpret for beginners.
* May be misleading if sample size is small.

images (1).png

```python
sns.violinplot(x=df['body_mass_g'],y=df['island'])
plt.show()
```

## PAIR PLOT


*   used for visualizing relationships between multiple variables in a dataset.

* By creating a grid of scatter plots it helps to identify how different features interact with each other to identify patterns, correlations and trends in data.


```python
sns.pairplot(data = df,height=1.5,hue='sex',palette='summer',diag_kind='hist')
plt.show()
```
