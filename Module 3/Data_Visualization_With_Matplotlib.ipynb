### BACKEND LAYER
Resposible for rendering visualizations and handle user interactions, it uses three components to do this which are following:


```
FigureCanvas
```
* It's acts as a drawing surface for the visual elements of the plot
* It is the area where plot is drawn and displayed

```
Renderer
```



*   It takes care of all the drawing operations on the Figure Canvas.
* it works as a painter that draws shapes,lines and text on the canvas

```
Event Handler System
```
  
*   The Event Handling System in Matplotlib is responsible for capturing and processing user interactions with the plot, such as mouse clicks, key presses, and other input events.



### ARTIST LAYER
In Matplotlib, an “artist” is any object that can be drawn to the canvas to create the visual representation of the data. Examples of artists include Figure, Axes, Axis, Line2D, Text, etc.


Here’s a brief explanation of these common artists:
```
Figure:
The top-level container that holds all the plot elements.
Axes:
The rectangular area within the Figure where data is plotted. It represents the coordinate system.
Axis:
The x-axis and y-axis of the plot, which contain ticks, tick labels, and axis labels.
Line2D:
Represents a line in the plot.
Text:
Represents text elements that can be added to the plot.

## SCRIPTING LAYER

* It is also known as pyplot
* pyplot is inspired by MATLAB's plotting interface
```
Feature and characteristics of scripting layer:
```
* State-based Interface: pyplot is a state-based interface, which means it keeps track of the current figure and axes internally.
* Automated Figure and Axes Creation: When you use pyplot, Figure and Axes objects are automatically created for you if they do not already exist. For instance, calling plt.plot() will create a Figure if it doesn't exist and then create an Axes within the Figure. If a Figure already exists, the plotting functions will use the existing Figure and Axes by default.

* Easy Plot Customization: pyplot provides a wide range of functions to customize your plots quickly
for example,plt.xlabel(), plt.title(), plt.legend()

### 3.1 LINE PLOT



*  Bivariate Analysis
*   Numerical vs Numerical

* Numerical vs Categorical

* Time Series Data - Data that changes with the time ,It always contain the timestamp and order

````python
price = [48000,54000,57000,49000,47000,45000]
year = [2015,2016,2017,2018,2019,2020]

plt.plot(year,price) # x = categorical data

# ylim
plt.ylim(0,100000)

# xlim

plt.xlim(2018,2020)
````

````python
price = [48000,54000,57000,49000,47000,45000]
year = [2015,2016,2017,2018,2019,2020]

plt.plot(year,price)

# giving the labels to the coordinates on the plot

for i in range(len(year)):
  plt.text(year[i],price[i],str([year[i],price[i]]),color='red',va = 'center',ha='center')
````

````python
from matplotlib.lines import lineStyles
df = pd.read_csv('/content/sharma-kohli.csv')
df

plt.plot(df['index'],df['V Kohli'],color='green',linestyle='solid',linewidth=4,marker='D',markersize=10)

plt.plot(df['index'],df['RG Sharma'],color='red',linestyle='dashdot',linewidth=5,marker='+',markersize = 12)

#title

plt.title('Rohit Sharma VS Virat Kholi Comparison')

#xlabel
plt.xlabel('Season')

# ylabel
plt.ylabel('Runs')

#legend
plt.legend(['Rohit Sharma','RG Sharma'],loc='lower right')

# grid
plt.grid()

# show()
plt.show()
````

````python
# histogram with meadian

data = [45,78,66,59,77,95,88,44,64,84,68,29,47]

# median
median_val = np.median(data)

# histogram

plt.hist(data,bins=5,color='skyblue', edgecolor='black', alpha=0.7)

# median line on x axis

plt.axvline(median_val,color='red',linestyle='dashed',linewidth=2)

# xlable
plt.xlabel('Student Marks')

# ylabel

plt.ylabel('Frequency')

# title

plt.title('Student Marks Histogram')

# legend
plt.legend(['Histogram','Median Line'])

# show

plt.show()
````

## Pie Chart


*   Univariate/Bivariate Analysis

* Categorical Vs Numerical Data

* Use Case - To find the contribution on a standard scale



````python
sub = ['hindi','english','mathmatics','sst','science']

score = [40,60,65,85,66]
````

````python
plt.pie(score,labels=sub,colors=['red','green','skyblue','yellow'],autopct='%1.3f%%',startangle=80,explode=(0,0,0,0,0.5),shadow=True)

plt.title('10th result')

# legend

plt.legend(loc='lower right')
plt.show()
````

### WHEN TO USE WHICH CHART
```
LINE CHART:

Use for: Continuous data changing over time or sequential intervals.
Data requirement: X-axis = Time/Sequence (Ordered), Y-axis = Numerical value.
Goal: Show trends, growth, or drops over a period.

BAR CHART:

Use for: Comparing discrete items, groups, or distinct categories.
Data requirement: X-axis = Categories, Y-axis = Numerical value.
Goal: Rank items or compare sizes across groups.
Rule of thumb: Keep categories under 7 for vertical bars; use horizontal bars if labels are long.

SCATTER PLOT:

Use for: Finding the correlation or relationship between two numerical variables.
Data requirement: X-axis = Numerical, Y-axis = Numerical.
Goal: Spot clusters, outliers, or positive/negative relationships.

HISTOGRAM:

Use for: Showing the frequency distribution of a single numerical variable.
Data requirement: X-axis = Bins/Intervals of a single continuous variable, Y-axis = Count/Frequency.
Goal: See how data is spread out (e.g., bell curves, skewed data).

PIE CHART:

Use for: Showing parts of a whole.
Data requirement: One categorical variable broken into parts adding up to 100%.
Goal: Show simple proportional breakdown.
Golden sub-rule: Never use more than 4–5 slices; use a bar chart instead if precise comparison is needed

```
